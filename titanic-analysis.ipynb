```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

def explore_data(df):
    print("SHAPE:", df.shape)
    print("\nMISSING VALUES:\n", df.isnull().sum())
    print("\nDUPLICATES:", df.duplicated().sum())
    print("\nDATA TYPES:\n", df.dtypes)
    print("\nSUMMARY:\n", df.describe())

explore_data(df)
```

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

print(df.isnull().sum())
```

```python
# Fix Age — fill with mean
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fix Cabin — drop the column
df = df.drop(columns=["Cabin"])

# Fix Embarked — fill with mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Check — should all be zero now
print(df.isnull().sum())
```

```python
print("Duplicates:", df.duplicated().sum())
```

```python
# Convert Sex to numbers — female=1, male=0
df["Sex"] = df["Sex"].map({"female": 1, "male": 0})

# Check it worked
print(df["Sex"].value_counts())
print(df["Sex"].dtype)
```

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv")

df["Age"] = df["Age"].fillna(df["Age"].mean())
df = df.drop(columns=["Cabin"])
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
df["Sex"] = df["Sex"].map({"female": 1, "male": 0})
```

```python
# Average age of survivors vs died
print(df.groupby("Survived")["Age"].mean())
```

```python
# Average fare of survivors vs died
print(df.groupby("Survived")["Fare"].mean())
```

```python
# Survival rate by class AND gender combined
print(df.groupby(["Pclass", "Sex"])["Survived"].mean())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Bar chart of survival count
sns.countplot(x="Survived", data=df)
plt.title("Survived vs Died")
plt.xlabel("0 = Died, 1 = Survived")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
sns.countplot(x="Survived", hue="Sex", data=df)
plt.title("Survival by Gender")
plt.xlabel("0 = Died, 1 = Survived")
plt.ylabel("Number of Passengers")
plt.legend(["Male", "Female"])
plt.show()
```

```python
sns.countplot(x="Pclass", hue="Survived", data=df)
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(["Died", "Survived"])
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df["Age"], bins=30, kde=True)
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.histplot(df["Fare"], bins=30, kde=True)
plt.title("Fare Distribution")
plt.xlabel("Fare (£)")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
